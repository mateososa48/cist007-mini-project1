### **Dataset Description**

Each row represents a penguin, and columns include:

- species – Penguin species
- island – Island where the penguin was found
- bill_length_mm – Bill length (mm)
- bill_depth_mm – Bill depth (mm)
- flipper_length_mm – Flipper length (mm)
- body_mass_g – Body mass (grams)
- sex – Sex of the penguin

Some values are missing, which makes this dataset ideal for practicing missing data handling.

Q20. Which variable is most strongly correlated with body_mass_g?

```python
numeric_columns = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'bill_ratio', 'body_mass_g']
mass_correlations = df[numeric_columns].corr()['body_mass_g'].drop('body_mass_g')
display(mass_correlations)
strongest = mass_correlations.abs().idxmax()
print('strongest:', strongest, round(mass_correlations[strongest], 2))
```

Q24. Replace missing values in the sex column with "Unknown".

```python
df['sex'] = df['sex'].replace('.', pd.NA).fillna('Unknown')
print('the . placeholder is included as Unknown.')
display(df['sex'].value_counts(dropna=False))
```

Q25. Fill missing values in body_mass_g using the mean of that column.

```python
mean_mass = df['body_mass_g'].mean()
df['body_mass_g'] = df['body_mass_g'].fillna(mean_mass)
print('mean used:', round(mean_mass, 2))
print('missing body masses left:', df['body_mass_g'].isna().sum())
```
