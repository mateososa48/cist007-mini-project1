### **Part 3 — Combining DataFrames + GroupBy**

Q12. Create the following penguin_regions dataframe and merge with penguins.



```
{
    'island': ['Biscoe', 'Dream', 'Torgersen'],
    'region': ['Antarctica', 'Antarctica', 'Antarctica']
}
```



````python
penguin_regions = pd.DataFrame({'island': ['Biscoe', 'Dream', 'Torgersen'], 'region': ['Antarctica', 'Antarctica', 'Antarctica']})
penguins_with_region = penguins.merge(penguin_regions, on='island', how='left')
display(penguins_with_region.head())
print('rows after merge:', len(penguins_with_region))
````

Q13. Compute the average body mass per region.

````python
display(penguins_with_region.groupby('region')['body_mass_g'].mean())
````

Q19. Convert observation_date to datetime and extract year.

````python
penguin_observations['observation_date'] = pd.to_datetime(penguin_observations['observation_date'])
penguin_observations['year'] = penguin_observations['observation_date'].dt.year
display(penguin_observations)
````

Q20. Merge with original penguins dataframe and show the head.

````python
# one-based IDs match the four example observations above
penguins_with_ids = penguins.copy()
penguins_with_ids['penguin_id'] = penguins_with_ids.index + 1
penguins_with_dates = penguins_with_ids.merge(penguin_observations, on='penguin_id', how='left')
display(penguins_with_dates.head())
print('rows after merge:', len(penguins_with_dates))
print('rows with example dates:', penguins_with_dates['observation_date'].notna().sum())
````
